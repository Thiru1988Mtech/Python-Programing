## Step 1: Create Sets

A **set** uses curly brackets. Duplicates are dropped and there is no order. `set()` makes an empty set, because `{}` makes an empty dictionary. `set(list)` removes duplicates from a list.

**Step 1: type this.** Make `tags` as `{"python", "code", "python"}`. Print `len(tags)`. Make `unique` from `set([3, 1, 3, 2, 1])` and print `sorted(unique)`. Print `type({})` and `type(set())`.

```python
tags = {"python", "code", "python"}
print(len(tags))
unique = set([3, 1, 3, 2, 1])
print(sorted(unique))
print(type({}), type(set()))
```

## Step 2: Set Operations

`|` is **union**, `&` is **intersection**, `-` is **difference** and `^` is **symmetric difference** (in one but not both). Sets have no order, so we print `sorted()` copies for stable output.

**Step 2: type this.** Set `a` to `{1, 2, 3}` and `b` to `{2, 3, 4}`. Print `sorted(a | b)`, `sorted(a & b)`, `sorted(a - b)` and `sorted(a ^ b)`. Then print `a <= {1, 2, 3, 4}` and `2 in a`.

```python
a = {1, 2, 3}
b = {2, 3, 4}
print(sorted(a | b))
print(sorted(a & b))
print(sorted(a - b))
print(sorted(a ^ b))
print(a <= {1, 2, 3, 4})
print(2 in a)
```

## Step 3: Add and Remove Set Items

`.add(x)` adds one item. `.update(list)` adds many. `.discard(x)` removes quietly. `.remove(x)` raises `KeyError` if the item is missing. Adding something already inside changes nothing.

**Step 3: type this.** Start with `colors = {"red"}`. `add("blue")`, `add("red")`, `update(["green", "pink"])`. Print `sorted(colors)`. Then `discard("pink")`, `discard("zzz")`, and print. Try `remove("zzz")` inside `try`.

```python
colors = {"red"}
colors.add("blue")
colors.add("red")
colors.update(["green", "pink"])
print(sorted(colors))
colors.discard("pink")
colors.discard("zzz")
print(sorted(colors))
try:
    colors.remove("zzz")
except KeyError as error:
    print("KeyError:", error)
```

## Step 4: Create Dictionaries

A **dictionary** has `key: value` pairs. Keys are unique: a repeated key keeps the **last** value. `dict()` also builds one. `len()` counts the pairs.

**Step 4: type this.** Make `user` with `name`, `age` and `city` for Asha, 21 and Chennai. Make `empty` as `{}`. Make `other` using `dict(name="Ravi", age=30)`. Print them, their `len` and `type`. Then `{"a": 1, "a": 2}`.

```python
user = {"name": "Asha", "age": 21, "city": "Chennai"}
empty = {}
other = dict(name="Ravi", age=30)
print(user)
print(empty, other)
print(len(user), type(user))
print({"a": 1, "a": 2})
```

## Step 5: Read Values

`d[key]` reads a value and raises `KeyError` if the key is missing. `d.get(key)` gives `None`, or a default that you choose. `key in d` checks the keys.

**Step 5: type this.** Use `user` from before. Print `user["name"]`, `user.get("age")`, `user.get("zip")`, `user.get("zip", "none")`, `"city" in user` and `"Asha" in user`. Try `user["zip"]` inside `try`.

```python
user = {"name": "Asha", "age": 21, "city": "Chennai"}
print(user["name"])
print(user.get("age"))
print(user.get("zip"))
print(user.get("zip", "none"))
print("city" in user, "Asha" in user)
try:
    print(user["zip"])
except KeyError as error:
    print("KeyError:", error)
```

## Step 6: Add and Update Entries

Assigning to a key **adds** it if it is new, and **updates** it if it exists. `.update(other)` merges many entries. `.setdefault(k, v)` adds only if the key is missing.

**Step 6: type this.** Make `user = {"name": "Asha"}`. Add `age` as 21. Update `age` to 22. Use `update` with `{"city": "Chennai", "age": 23}`. Use `setdefault("age", 99)` and `setdefault("zip", 600001)`. Print `user` at the end.

```python
user = {"name": "Asha"}
user["age"] = 21
user["age"] = 22
print(user)
user.update({"city": "Chennai", "age": 23})
print(user)
user.setdefault("age", 99)
user.setdefault("zip", 600001)
print(user)
```

## Step 7: Remove Entries

`.pop(k)` removes a key and returns its value. `.pop(k, default)` is safe. `del d[k]` removes without returning. `.clear()` empties the dictionary.

**Step 7: type this.** Make `d` with `a`, `b`, `c`, `e` as 1, 2, 3, 5. Print `d.pop("b")`, `d.pop("zzz", "missing")`. Then `del d["a"]` and print `d`. Finally `d.clear()` and print `d`.

```python
d = {"a": 1, "b": 2, "c": 3, "e": 5}
print(d.pop("b"))
print(d.pop("zzz", "missing"))
del d["a"]
print(d)
d.clear()
print(d)
```

## Step 8: Keys, Values and Items

`.keys()`, `.values()` and `.items()` give views of the dictionary. Wrap them in `list()` to see them. `.items()` gives `(key, value)` tuples. Keys must be **immutable**, such as text, numbers and tuples.

**Step 8: type this.** Make `prices = {"tea": 10, "cake": 40}`. Print `list(prices.keys())`, `list(prices.values())` and `list(prices.items())`. Print `sum(prices.values())`. Make `grid = {(0, 0): "start"}` and print `grid[(0, 0)]`. Try a list as a key inside `try`.

```python
prices = {"tea": 10, "cake": 40}
print(list(prices.keys()))
print(list(prices.values()))
print(list(prices.items()))
print(sum(prices.values()))
grid = {(0, 0): "start"}
print(grid[(0, 0)])
try:
    bad = {[0, 0]: "no"}
except TypeError as error:
    print("TypeError:", error)
```

## Step 9: Mini Practice: Phone Book

Let's build a small phone book. We add contacts, update one, look up with a safe default, remove one and count the rest.

**Step 9: type this.** Make `book` as `{"Asha": "98400"}`. Add `Ravi`. Update `Asha`'s number. Look up `Meena` with `get` and a default of `"not found"`. Remove `Ravi`. Print `book` and `len(book)`.

```python
book = {"Asha": "98400"}
book["Ravi"] = "98410"
book["Asha"] = "98499"
print(book.get("Meena", "not found"))
removed = book.pop("Ravi")
print(removed)
print(book, len(book))
```

## Recap

- A set keeps **unique** values, with no order. `set()` is the empty set. `|`, `&`, `-`, `^` combine sets.
- `add`, `update`, `discard` and `remove` change a set. `discard` is quiet, `remove` raises `KeyError`.
- A dictionary stores `key: value` pairs. Keys are unique and immutable.
- `d[k]` is strict. `d.get(k, default)` is safe. `k in d` checks keys.
- Add or update with `d[k] = v`, `update`, `setdefault`. Remove with `pop`, `del`, `clear`.
