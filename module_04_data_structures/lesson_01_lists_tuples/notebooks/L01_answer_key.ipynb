## Step 1: Create Lists

A **list** is written with square brackets. Items are separated by commas, and can be of any type. `[]` is an empty list. `len()` counts the items.

**Step 1: type this.** Make `marks` with `80, 95, 70`, `mixed` with `"Asha", 21, True`, and an empty list `empty`. Print all three, their `len`, and `type(marks)`.

```python
marks = [80, 95, 70]
mixed = ["Asha", 21, True]
empty = []
print(marks, mixed, empty)
print(len(marks), len(mixed), len(empty))
print(type(marks))
```

## Step 2: Index and Slice a List

Indexes start at **0**, and negative indexes count from the end. A slice gives a **new list**. It works exactly as it does for strings.

**Step 2: type this.** Set `nums` to `[30, 10, 20, 40, 50]`. Print `nums[0]`, `nums[-1]`, `nums[1:3]`, `nums[:2]`, `nums[::-1]` and `nums[::2]`.

```python
nums = [30, 10, 20, 40, 50]
print(nums[0])
print(nums[-1])
print(nums[1:3])
print(nums[:2])
print(nums[::-1])
print(nums[::2])
```

## Step 3: Change Items

A list is **mutable**: assign to an index to change an item, or assign to a slice to change several at once. Lists also allow `+` to join and `*` to repeat.

**Step 3: type this.** Set `nums` to `[30, 10, 20]`. Change `nums[0]` to `5`. Change `nums[1:]` to `[7, 8, 9]`. Print `nums`. Then print `[1, 2] + [3]` and `[0] * 3`.

```python
nums = [30, 10, 20]
nums[0] = 5
print(nums)
nums[1:] = [7, 8, 9]
print(nums)
print([1, 2] + [3])
print([0] * 3)
```

## Step 4: Add Items

`.append(x)` adds one item at the end. `.insert(i, x)` adds at index `i`. `.extend(list)` adds many items. Note the difference: `append([1, 2])` adds **one** list as one item.

**Step 4: type this.** Start with `items = ["a"]`. Use `append("b")`, `insert(0, "z")` and `extend(["c", "d"])`, printing after each. Then `append(["e", "f"])` and print.

```python
items = ["a"]
items.append("b")
print(items)
items.insert(0, "z")
print(items)
items.extend(["c", "d"])
print(items)
items.append(["e", "f"])
print(items)
```

## Step 5: Remove Items

`.remove(x)` deletes the first item equal to `x`. `.pop()` removes the last item and returns it. `.pop(i)` removes index `i`. `del items[i]` also deletes. `.clear()` empties the list.

**Step 5: type this.** Set `items` to `["a", "b", "c", "d", "e"]`. Use `remove("b")`, `pop()`, `pop(0)` and `del items[0]`, printing what pop returns and the list each time. Then `clear()`.

```python
items = ["a", "b", "c", "d", "e"]
items.remove("b")
print(items)
last = items.pop()
print(last, items)
first = items.pop(0)
print(first, items)
del items[0]
print(items)
items.clear()
print(items)
```

## Step 6: Search and Count

`x in items` checks membership. `.index(x)` gives the position. `.count(x)` counts matches. `min()`, `max()` and `sum()` work on lists of numbers. `.index()` raises `ValueError` when the item is missing.

**Step 6: type this.** Set `nums` to `[4, 8, 4, 2]`. Print `4 in nums`, `nums.index(8)`, `nums.count(4)`, `min`, `max` and `sum`. Then try `nums.index(99)` inside `try` and print the error.

```python
nums = [4, 8, 4, 2]
print(4 in nums)
print(nums.index(8))
print(nums.count(4))
print(min(nums), max(nums), sum(nums))
try:
    nums.index(99)
except ValueError as error:
    print("ValueError:", error)
```

## Step 7: Tuples

A **tuple** uses round brackets. Reading works like a list. Changing raises `TypeError`. A one-item tuple needs a trailing comma. You can build a tuple without brackets too.

**Step 7: type this.** Make `point` as `(3, 4)`. Print `point[0]`, `point[-1]` and `point[:1]`. Make `one` as `(5,)` and `not_tuple` as `(5)` and print their types. Try `point[0] = 9` inside `try`.

```python
point = (3, 4)
print(point[0], point[-1], point[:1])
one = (5,)
not_tuple = (5)
print(type(one), type(not_tuple))
print((3, 4) == point)
try:
    point[0] = 9
except TypeError as error:
    print("TypeError:", error)
```

## Step 8: Sort with sorted() and .sort()

`sorted(x)` returns a **new** list. `.sort()` changes the list **in place** and returns `None`. Add `reverse=True` for descending order.

**Step 8: type this.** Set `nums` to `[30, 10, 20]`. Print `sorted(nums)` and `nums`. Then call `nums.sort()` and print `nums`. Then `nums.sort(reverse=True)` and print. Last, `result = nums.sort()` and print `result`.

```python
nums = [30, 10, 20]
print(sorted(nums))
print(nums)
nums.sort()
print(nums)
nums.sort(reverse=True)
print(nums)
result = nums.sort()
print(result)
```

## Step 9: Sort with a Key

`key=` tells Python **what to sort by**. `key=len` sorts words by length. `key=str.lower` ignores capital letters. Sorting a list of words also works for text in alphabetical order.

**Step 9: type this.** Set `words` to `["pear", "Fig", "apple", "kiwi"]`. Print `sorted(words)`, `sorted(words, key=str.lower)`, `sorted(words, key=len)` and `sorted(words, key=len, reverse=True)`.

```python
words = ["pear", "Fig", "apple", "kiwi"]
print(sorted(words))
print(sorted(words, key=str.lower))
print(sorted(words, key=len))
print(sorted(words, key=len, reverse=True))
```

## Step 10: Mini Practice

Let's use everything. Given a list of marks, find the best three, the lowest mark, and the average. Sorting, slicing, `min`, `sum` and `len` all work together.

**Step 10: type this.** Set `marks` to `[72, 95, 60, 88, 79]`. Make `top3` using `sorted(..., reverse=True)` and a slice. Print `top3`, `min(marks)` and the average with `round(..., 1)`.

```python
marks = [72, 95, 60, 88, 79]
top3 = sorted(marks, reverse=True)[:3]
print(top3)
print(min(marks))
average = sum(marks) / len(marks)
print(round(average, 1))
```

## Recap

- A list is ordered and **changeable**: `[1, 2, 3]`. Index, slice and `len()` work like strings.
- Add with `append`, `insert`, `extend`. Remove with `remove`, `pop`, `del`, `clear`.
- A tuple is `(1, 2)`: ordered and **fixed**. A one-item tuple needs a comma.
- `sorted()` makes a new list. `.sort()` works in place and returns `None`.
- `key=` sorts by a rule. `reverse=True` goes descending.
