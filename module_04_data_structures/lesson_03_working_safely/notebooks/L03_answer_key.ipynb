## Step 1: Two Names, One List

`b = a` does **not** copy. Both names point to the same list. `is` checks whether two names point to the same object.

**Step 1: type this.** Set `a` to `[1, 2, 3]` and `b = a`. Append `4` through `b`. Print `a`, `a is b` and the `id` check `id(a) == id(b)`.

```python
a = [1, 2, 3]
b = a
b.append(4)
print(a)
print(a is b)
print(id(a) == id(b))
```

## Step 2: Make Real Copies

Three easy ways to copy a flat list: `.copy()`, `list(a)` and `a[:]`. Each makes a **new** list. Dictionaries and sets also have `.copy()`.

**Step 2: type this.** Set `a` to `[1, 2, 3]`. Make `c1`, `c2`, `c3` with `.copy()`, `list(a)` and `a[:]`. Append `9` to `c1` only. Print `a`, `c1`, `c2`, `a is c1`. Copy a dictionary with `.copy()` and change the copy.

```python
a = [1, 2, 3]
c1 = a.copy()
c2 = list(a)
c3 = a[:]
c1.append(9)
print(a, c1)
print(c2 == c3, c2 is c3)
print(a is c1)
d = {"x": 1}
e = d.copy()
e["x"] = 99
print(d, e)
```

## Step 3: Shallow vs Deep Copy

`.copy()` is **shallow**: inner lists are still shared. `copy.deepcopy()` copies every level. Use it for nested data.

**Step 3: type this.** Import `copy`. Set `x` to `[[1], [2]]`. Make `y = x.copy()` and append `9` to `y[0]`. Print `x`. Then make `z = copy.deepcopy(x)`, append `7` to `z[1]`, and print `x` and `z`.

```python
import copy
x = [[1], [2]]
y = x.copy()
y[0].append(9)
print(x)
z = copy.deepcopy(x)
z[1].append(7)
print(x)
print(z)
```

## Step 4: Unpack Lists and Tuples

**Unpacking** assigns each item to a name in one line. The number of names must match the number of items. A mismatch raises `ValueError`. It works on lists, tuples and strings.

**Step 4: type this.** Unpack `(3, 4)` into `x, y`. Unpack `"abc"` into `p, q, r`. Print them. Then try `m, n = [1, 2, 3]` inside `try` and print the error.

```python
x, y = (3, 4)
p, q, r = "abc"
print(x, y)
print(p, q, r)
try:
    m, n = [1, 2, 3]
except ValueError as error:
    print("ValueError:", error)
```

## Step 5: Swap and Star

`a, b = b, a` swaps two values with no helper variable. A name with `*` collects the rest as a **list**. `_` is the usual name for a value you do not need.

**Step 5: type this.** Set `a, b = 1, 2` and swap them. Print `a, b`. Unpack `[10, 20, 30, 40]` into `first, *middle, last`. Print all three. Unpack `(1, 2)` into `_, second` and print `second`.

```python
a, b = 1, 2
a, b = b, a
print(a, b)
first, *middle, last = [10, 20, 30, 40]
print(first, middle, last)
_, second = (1, 2)
print(second)
```

## Step 6: Mutable and Immutable

Numbers, strings and tuples are **immutable**. Lists, dictionaries and sets are **mutable**. `x += 1` on a number makes a **new** number. `+=` on a list changes the list **in place**.

**Step 6: type this.** Set `n = 5`, `m = n`, then `n += 1`. Print `n` and `m`. Set `a = [1]`, `b = a`, then `a += [2]`. Print `b`. Set `t = (1, [2])`, append `3` to `t[1]` and print `t`.

```python
n = 5
m = n
n += 1
print(n, m)
a = [1]
b = a
a += [2]
print(b)
t = (1, [2])
t[1].append(3)
print(t)
```

## Step 7: Nested Lists

A list can hold lists. Chain indexes to reach inside: `grid[1][2]` means the row at index 1, then the item at index 2. Make rows with a loop or a comprehension, **not** with `[[0] * 2] * 3`.

**Step 7: type this.** Set `grid` to `[[1, 2, 3], [4, 5, 6]]`. Print `grid[1][2]`, `grid[0]`, `len(grid)`. Set `grid[0][0] = 99` and print `grid`. Then make `bad = [[0] * 2] * 2`, set `bad[0][0] = 1` and print `bad`.

```python
grid = [[1, 2, 3], [4, 5, 6]]
print(grid[1][2], grid[0], len(grid))
grid[0][0] = 99
print(grid)
bad = [[0] * 2] * 2
bad[0][0] = 1
print(bad)
```

## Step 8: Nested Dictionaries

Dictionaries can hold lists and dictionaries. Read step by step. Use `.get()` with a default so that a missing key does not stop the program.

**Step 8: type this.** Make `shop` with `tea` as a list `[10, 12]` and `cake` as a dictionary `{"price": 40}`. Print `shop["tea"][1]` and `shop["cake"]["price"]`. Append `15` to tea. Add `stock` to cake. Print `shop`. Then print `shop.get("juice", {}).get("price", 0)`.

```python
shop = {"tea": [10, 12], "cake": {"price": 40}}
print(shop["tea"][1])
print(shop["cake"]["price"])
shop["tea"].append(15)
shop["cake"]["stock"] = 5
print(shop)
print(shop.get("juice", {}).get("price", 0))
```

## Step 9: Unpack a Dictionary

`.items()` gives pairs, and each pair can be unpacked into two names. `dict(zip(keys, values))` builds a dictionary from two lists. `{**a, **b}` merges two dictionaries into a new one.

**Step 9: type this.** Set `prices = {"tea": 10, "cake": 40}`. Take `first = list(prices.items())[0]` and unpack it into `item, cost`. Print them. Make `d` from `zip(["a", "b"], [1, 2])`. Merge `{"a": 1}` and `{"a": 5, "c": 3}` with `{**x, **y}`.

```python
prices = {"tea": 10, "cake": 40}
item, cost = list(prices.items())[0]
print(item, cost)
d = dict(zip(["a", "b"], [1, 2]))
print(d)
x = {"a": 1}
y = {"a": 5, "c": 3}
print({**x, **y})
print(x)
```

## Step 10: Mini Practice: Gradebook

Let's use everything. A gradebook is a dictionary of students, and each student is a dictionary with a list of marks. We copy it safely, change the copy and compute an average.

**Step 10: type this.** Make `book` with `Asha` and `Ravi`, each with `{"marks": [80, 90]}` and `{"marks": [70, 60]}`. Make `backup = copy.deepcopy(book)`. Add `100` to Asha's marks in `book`. Print both Asha lists. Then print Asha's average using `sum` and `len`.

```python
import copy
book = {"Asha": {"marks": [80, 90]}, "Ravi": {"marks": [70, 60]}}
backup = copy.deepcopy(book)
book["Asha"]["marks"].append(100)
print(book["Asha"]["marks"])
print(backup["Asha"]["marks"])
marks = book["Asha"]["marks"]
print(sum(marks) / len(marks))
```

## Recap

- `b = a` shares one object. Use `.copy()` (shallow) or `copy.deepcopy()` (nested) for a real copy.
- Unpack with `x, y = pair`. Swap with `a, b = b, a`. `*rest` collects extra items as a list.
- Immutable: numbers, strings, tuples. Mutable: lists, dictionaries, sets.
- Reach nested data by chaining: `shop["cake"]["price"]`. Use `.get()` with defaults for safety.
- `[[0] * 2] * 2` shares the inner lists. Merge dictionaries with `{**a, **b}`.
