## Step 1: Create Strings

A **string** is text inside quotes. Single and double quotes work the same way. The empty string `""` is a valid string with nothing inside.

**Step 1: type this.** Make `name` with `Asha`, `city` with `Chennai` (use single quotes) and `empty` with empty text. Print `name` and `city`. Then print the types of `name` and `empty`.

```python
name = "Asha"
city = 'Chennai'
empty = ""
print(name, city)
print(type(name), type(empty))
```

## Step 2: Quotes Inside Strings

If your text contains a quote, wrap it in the **other** kind of quote. Triple quotes can hold text that goes over several lines.

**Step 2: type this.** Print `It's easy`, then print `He said "hi"`, and then print a two-line text in triple quotes.

```python
print("It's easy")
print('He said "hi"')
print("""Line one
Line two""")
```

## Step 3: Length with len()

`len()` counts the characters in a string. **Every** character counts: letters, spaces and symbols. The empty string has length 0.

**Step 3: type this.** Set `word` to `python`. Print `len(word)`, `len("")`, `len("a b")` and `len("  ")` (two spaces).

```python
word = "python"
print(len(word))
print(len(""))
print(len("a b"))
print(len("  "))
```

## Step 4: Indexing from the Front

An **index** is the position of a character. Counting starts at **0**. Put the index in square brackets. The answer is a string with one character.

**Step 4: type this.** Set `word` to `python`. Print `word[0]`, `word[1]` and `word[5]`.

```python
word = "python"
print(word[0])
print(word[1])
print(word[5])
```

## Step 5: Negative Indexing

A **negative index** counts from the end. `-1` is the last character, `-2` the one before it. `word[-1]` is the same as `word[len(word) - 1]`.

**Step 5: type this.** Set `word` to `python`. Print `word[-1]`, `word[-2]`, `word[-6]` and `word[len(word) - 1]`.

```python
word = "python"
print(word[-1])
print(word[-2])
print(word[-6])
print(word[len(word) - 1])
```

## Step 6: Index Errors

An index that is too big raises an `IndexError`. We catch it with `try` and `except` so that the notebook keeps running. You will study them in Module 09.

**Step 6: type this.** Try `word[6]` and print the message of the `IndexError`. Then print `word[5]` to show the largest valid index.

```python
word = "python"
try:
    print(word[6])
except IndexError as error:
    print("IndexError:", error)
print(word[5])
```

## Step 7: Slicing Basics

A slice is `text[start:stop]`. The start is **included** and the stop is **not**. If you leave out the start, Python begins at the front. If you leave out the stop, Python goes to the end.

**Step 7: type this.** Set `word` to `python`. Print `word[1:4]`, `word[:3]`, `word[3:]`, `word[:]` and `word[-3:]`.

```python
word = "python"
print(word[1:4])
print(word[:3])
print(word[3:])
print(word[:])
print(word[-3:])
```

## Step 8: Slicing with a Step

A third number is the **step**. `[::2]` takes every second character. A negative step walks backwards, so `[::-1]` reverses a string.

**Step 8: type this.** Set `word` to `python`. Print `word[::2]`, `word[1::2]`, `word[::-1]` and `word[4:1:-1]`.

```python
word = "python"
print(word[::2])
print(word[1::2])
print(word[::-1])
print(word[4:1:-1])
```

## Step 9: Slices Never Fail

Unlike indexing, a slice never raises an error for numbers that are too big. You simply get a shorter string, or an **empty** string. `repr()` shows the quotes, so an empty string is easy to see.

**Step 9: type this.** Set `word` to `python`. Print `word[2:100]`. Print `repr(word[10:])` and `repr(word[4:2])`.

```python
word = "python"
print(word[2:100])
print(repr(word[10:]))
print(repr(word[4:2]))
```

## Step 10: Mini Practice

Let's use everything. A **palindrome** reads the same forwards and backwards, like `level`. We take the first letter, the last letter, the middle part, and compare the word with its reverse.

**Step 10: type this.** Set `word` to `level`. Make `first`, `last` and `middle` (without the first and last letters). Print them. Then print whether `word == word[::-1]`.

```python
word = "level"
first = word[0]
last = word[-1]
middle = word[1:-1]
print(first, last, middle)
print(word == word[::-1])
```

## Recap

- A string is a row of characters in quotes. Use the other kind of quote inside, or triple quotes for many lines.
- `len()` counts every character, including spaces. The empty string has length 0.
- Index counts from **0**. Negative indexes count from the end: `word[-1]` is the last character.
- Too big an index raises `IndexError`. A slice never fails.
- `text[start:stop:step]`. Start is included, stop is not. `text[::-1]` reverses.
