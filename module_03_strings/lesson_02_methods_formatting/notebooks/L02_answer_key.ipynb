## Step 1: Join Strings

The `+` operator **joins** strings. This is called concatenation. Python adds nothing between them, so add your own spaces.

**Step 1: type this.** Make `first` as `Asha` and `last` as `Kumar`. Print `first + last`, then `first + " " + last`.

```python
first = "Asha"
last = "Kumar"
print(first + last)
print(first + " " + last)
```

## Step 2: Repeat and the str() Trap

`*` **repeats** a string. The `+` operator needs strings on both sides, so a number must go through `str()` first. We catch the error so the notebook keeps running.

**Step 2: type this.** Print `"ha" * 3` and `"-" * 10`. Then try `"age: " + 21` inside `try` and print the error. Finally print `"age: " + str(21)`.

```python
print("ha" * 3)
print("-" * 10)
try:
    print("age: " + 21)
except TypeError as error:
    print("TypeError:", error)
print("age: " + str(21))
```

## Step 3: Escapes and Raw Strings

A backslash starts an **escape sequence**: `\n` is a new line, `\t` is a tab, `\\` is one backslash. A **raw string**, `r"..."`, keeps backslashes as typed.

**Step 3: type this.** Print `"one\ntwo"`, then `"a\tb"`, then `"C:\\new"`. Print the raw string `r"C:\new"` and the `len` of both.

```python
print("one\ntwo")
print("a\tb")
print("C:\\new")
print(r"C:\new")
print(len("C:\\new"), len(r"C:\new"))
```

## Step 4: Case, Strip and Length

Methods are called with a dot. `.upper()` and `.lower()` change the case. `.strip()` removes spaces at both ends. A method returns a **new** string.

**Step 4: type this.** Set `text` to `"  Python Fun  "`. Print `repr(text)`, `repr(text.strip())`, `text.upper()`, `text.lower()` and the length before and after `strip()`.

```python
text = "  Python Fun  "
print(repr(text))
print(repr(text.strip()))
print(text.upper())
print(text.lower())
print(len(text), len(text.strip()))
```

## Step 5: Replace, Find and Count

`.replace(a, b)` swaps text. `.find(x)` gives the index of the first match, or `-1` if there is none. `.count(x)` counts matches. `.startswith()` and `.endswith()` give True or False.

**Step 5: type this.** Set `text` to `banana`. Print `text.replace("a", "o")`, `text.find("na")`, `text.find("z")`, `text.count("a")`, `text.startswith("ban")` and `text.endswith("x")`.

```python
text = "banana"
print(text.replace("a", "o"))
print(text.find("na"))
print(text.find("z"))
print(text.count("a"))
print(text.startswith("ban"))
print(text.endswith("x"))
```

## Step 6: Split and Join

`.split()` cuts a string into a **list** of pieces. `separator.join(list)` glues a list back into one string. They are opposites.

**Step 6: type this.** Set `sentence` to `"learn python step by step"`. Print `sentence.split()`. Make `words` from it. Print `"-".join(words)`. Print `"a,b,c".split(",")`.

```python
sentence = "learn python step by step"
print(sentence.split())
words = sentence.split()
print("-".join(words))
print("a,b,c".split(","))
```

## Step 7: Checks and Capitalize

Some methods answer a yes or no question: `.isdigit()`, `.isalpha()`, `.isupper()`. `.capitalize()` and `.title()` fix the case of names.

**Step 7: type this.** Print `"2024".isdigit()`, `"abc".isalpha()`, `"abc1".isalpha()` and `"ABC".isupper()`. Then print `"asha kumar".capitalize()` and `"asha kumar".title()`.

```python
print("2024".isdigit())
print("abc".isalpha())
print("abc1".isalpha())
print("ABC".isupper())
print("asha kumar".capitalize())
print("asha kumar".title())
```

## Step 8: Formatting Styles

Three ways to put values inside text. The `%` style and `.format()` are older. **f-strings** are the modern choice. They give the same result.

**Step 8: type this.** Set `name` to `Asha` and `age` to `21`. Print the same sentence three ways: with `%`, with `.format()` and with an f-string.

```python
name = "Asha"
age = 21
print("%s is %d" % (name, age))
print("{} is {}".format(name, age))
print(f"{name} is {age}")
```

## Step 9: f-string Format Specs

Inside `{}` you can write any expression, and after a colon a **format spec**. `.2f` is 2 decimals, `,` adds thousand commas, `.1%` makes a percentage, and `>6`, `<6`, `^6` align text. `{age=}` shows the name and the value.

**Step 9: type this.** Set `name`, `age`, `price` and `ratio`. Print `{age + 1}`, `{price:.2f}`, `{1234567:,}`, `{ratio:.1%}`, three aligned words in `|` bars, and `{age=}`.

```python
name = "Asha"
age = 21
price = 3.14159
ratio = 0.256
print(f"next year: {age + 1}")
print(f"{price:.2f}")
print(f"{1234567:,}")
print(f"{ratio:.1%}")
print(f"|{name:>6}|{name:<6}|{name:^6}|")
print(f"{age=}")
```

## Step 10: Strings Are Immutable

A string **cannot change** after it is created. A method gives a **new** string. Assigning to one character gives a `TypeError`. To change text, build a new string with slices and `+`.

**Step 10: type this.** Set `word` to `python`. Print `word.upper()` and then `word`. Try `word[0] = "P"` inside `try`. Then print `"P" + word[1:]`.

```python
word = "python"
print(word.upper())
print(word)
try:
    word[0] = "P"
except TypeError as error:
    print("TypeError:", error)
print("P" + word[1:])
```

## Recap

- `+` joins and `*` repeats. A number needs `str()` before `+`.
- A backslash starts an escape. `\n` new line, `\t` tab, `\\` backslash. `r"..."` keeps backslashes.
- Methods give **new** strings: `strip`, `lower`, `upper`, `replace`, `split`, `join`, `find`, `count`.
- f-strings: `f"{value:spec}"`. Use `.2f`, `,` and `.1%`.
- Strings are **immutable**. Build a new string, and assign it back to keep it.
