## Step 1: Check the Python Version

Python is already running this notebook. Let's ask it for its version. The built-in `sys` module knows facts about Python itself. We need **3.11 or newer** for this series.

**Step 1: type this.** Import the `sys` module. Print `sys.version`. Then print whether `sys.version_info` is at least `(3, 11)`.

```python
import sys
print(sys.version)
print("Python 3.11 or newer?", sys.version_info >= (3, 11))
```

## Step 2: Use Python as a Calculator

The interpreter can work like a calculator. It reads a line, works out the answer and shows it. `+` adds, `/` divides and `**` means *to the power of*.

**Step 2: type this.** Print the result of `2 + 3`, then `10 / 4`, then `2 ** 10`.

```python
print(2 + 3)
print(10 / 4)
print(2 ** 10)
```

## Step 3: Peek Inside the Interpreter

Remember the picture from the slides: source code becomes bytecode, and then Python runs it. Here we do those two steps by hand. `compile()` builds the code object and `exec()` runs it. You will not need this in daily work. It just shows what Python does for you.

**Step 3: type this.** Call `compile()` on the text `print('Hello from compiled code')`. Print the type of the result, then run it with `exec()`.

```python
code = compile("print('Hello from compiled code')", "<demo>", "exec")
print(type(code))
exec(code)
```

## Step 4: Write Your First Program

Here is the classic first program. `print()` is a built-in **function**, a small tool that does one job. Here, the job is to show text on the screen. The text goes inside quotes.

**Step 4: type this.** Use `print()` to show the text `Hello, World!`.

```python
print("Hello, World!")
```

## Step 5: Print More Than One Thing

`print()` can show several things in one line. Separate them with commas, and Python puts a space between them. Text needs quotes. **Numbers do not.** Python can also calculate inside the brackets.

**Step 5: type this.** Print the three words `Python`, `is` and `fun` in one `print()`. Then print the result of `7 * 6`.

```python
print("Python", "is", "fun")
print(7 * 6)
```

## Step 6: Add Comments

A **comment** starts with `#`. Python ignores everything after it on that line. Use comments to explain *why*. You can also put `#` before a line of code to switch it off for a while.

**Step 6: type this.** Write a full-line comment. Write a `print()` line with an inline comment after it. Then switch off a second `print()` line with `#`.

```python
# This is a comment. Python ignores it.
print("Comments do not run")   # inline comment
# print("This line is switched off")
```

## Step 7: Use Triple Quotes for Notes

Python has no special tool for long comments. Many people use a **triple-quoted string** for longer notes. It is real text, not a comment, but it does nothing when it stands alone. The same quotes are used for **docstrings**, which you will meet later in Module 06.

**Step 7: type this.** Write a note on three lines between triple quotes. Then add a `print()` line.

```python
"""
This is a triple-quoted string.
People use it for longer notes.
"""
print("Triple quotes can hold notes")
```

## Step 8: Read an Error Message

Everyone makes mistakes, so let's make one **on purpose**. Python tells us what went wrong. In real code, an error stops the program. Here we catch it with `try` and `except` so that the notebook keeps running. You will learn `try` and `except` in Module 09. For now, just read the message.

**Step 8: type this.** Run the code `print(Hello)` without quotes. Catch the `NameError` and print its message.

```python
try:
    exec("print(Hello)")
except NameError as error:
    print("NameError:", error)
```

## Recap

- Python is read by an **interpreter** that turns code into bytecode and runs it.
- Use Python **3.12** (minimum 3.11). Check it with `python --version` or `sys.version`.
- `print()` shows text and numbers. Text needs quotes, numbers do not.
- A comment starts with `#`. Triple quotes can hold longer notes.
- Error messages are friendly clues. Read the last line first.
