## Step 1: Control print()

`print()` has two helpful options. `sep` is what goes **between** the values (default: a space). `end` is what goes **after** the last value (default: a new line).

**Step 1: type this.** Print `Hello Python` with a space. Print it again with `sep="-"`. Then print `Loading` with `end="..."` followed by a normal `print("done")`.

```python
print("Hello", "Python")
print("Hello", "Python", sep="-")
print("Loading", end="...")
print("done")
```

## Step 2: Print Variables in a Sentence

You can mix text and variables. Commas work well. An **f-string** is a quick way to put variables inside text. You write `f` before the quote and put names inside `{}`. We study f-strings in Module 03. Here is a first taste.

**Step 2: type this.** Set `name` to `Asha` and `age` to `21`. Print them with a comma version, and then with an f-string.

```python
name = "Asha"
age = 21
print("Name:", name, "Age:", age)
print(f"{name} is {age} years old")
```

## Step 3: Read Input from the User

`input()` shows a prompt and waits. Whatever the user types comes back as **text** (a `str`). In this notebook, Jupyter shows a small box where you type your answer.

**Step 3: type this.** Ask for the user's name with the prompt `Enter your name: `. Print a greeting, and then print the type of the answer.

```python
name = input("Enter your name: ")
print("Hello,", name)
print(type(name))
```

## Step 4: Input Is Always Text

Here is the famous trap. If the user types `21`, `input()` gives you the **text** `"21"`. Adding `1` to the text only *joins* the texts. To do maths, convert with `int()` first.

**Step 4: type this.** Ask for the age. Convert it with `int()`. Print the age next year. Then print `age_text + "1"` to see the trap.

```python
age_text = input("Enter your age: ")
age = int(age_text)
print("Next year you will be", age + 1)
print("Without int():", age_text + "1")
```

## Step 5: Convert Between Types

`int()`, `float()` and `str()` convert values. Note that `int(3.9)` **cuts off** the decimals. It does not round.

**Step 5: type this.** Print `int("42") + 1`, `float("3.5") * 2`, `str(7) + "0"`, `int(3.9)` and `float(5)`.

```python
print(int("42") + 1)
print(float("3.5") * 2)
print(str(7) + "0")
print(int(3.9))
print(float(5))
```

## Step 6: When Conversion Fails

Not every text can become a number. `int("3.5")` fails with a `ValueError`. Go in two steps instead. Also, `bool("False")` is `True`, because the text is not empty. We use `try` and `except` only to show the error without stopping the notebook (Module 09 explains them).

**Step 6: type this.** Try `int("3.5")` and print the error message. Then print `int(float("3.5"))` and `bool("False")`.

```python
try:
    int("3.5")
except ValueError as error:
    print("ValueError:", error)
print(int(float("3.5")))
print(bool("False"))
```

## Step 7: Syntax and Indentation

A **colon** opens a block, and the lines in the block are **indented** by four spaces. Lines at the same level belong together. The last line has no indent, so it is outside the block.

**Step 7: type this.** Set `temperature` to 32. Write `if temperature > 30:` and put two `print()` lines under it, indented. Add a last `print()` that is not indented.

```python
temperature = 32
if temperature > 30:
    print("It is hot")
    print("Drink water")
print("This line always runs")
```

## Step 8: See an Indentation Error

Let's break the rule on purpose. We give Python a program in which the line after `if True:` is **not** indented. Python refuses to run it and raises an `IndentationError`.

**Step 8: type this.** Compile the text `if True:` followed by an unindented `print('hi')`. Catch the `IndentationError` and print its message.

```python
try:
    compile("if True:\nprint('hi')", "<demo>", "exec")
except IndentationError as error:
    print("IndentationError:", error.msg)
```

## Step 9: Truthy and Falsy

Any value can act as `True` or `False`. **Falsy** values are `0`, `0.0`, `""` (empty text), `None` and `False`. Almost everything else is **truthy**, even the text `"0"`. `bool()` shows how Python sees a value.

**Step 9: type this.** Print `bool()` of `0`, `0.0`, `""` and `None`. Then print `bool()` of `1`, `-5`, `"0"` and `" "`. Finally, set `name` to empty text and use `if name:` with an `else:` to print whether a name was given.

```python
print(bool(0), bool(0.0), bool(""), bool(None))
print(bool(1), bool(-5), bool("0"), bool(" "))
name = ""
if name:
    print("Name given")
else:
    print("Name is empty")
```

## Step 10: Use Built-in Functions

Built-ins are tools that Python gives you for free. `len()` counts, `abs()` removes the minus sign, `round()` rounds, `min()` and `max()` find the smallest and the largest, and `sum()` adds up a list. The square brackets make a **list**, which you will learn in Module 04.

**Step 10: type this.** Print `len("python")`, `abs(-7)`, `round(3.14159, 2)`, `min(4, 9, 2)` and `max(4, 9, 2)` and `sum([1, 2, 3, 4])`.

```python
print(len("python"))
print(abs(-7))
print(round(3.14159, 2))
print(min(4, 9, 2), max(4, 9, 2))
print(sum([1, 2, 3, 4]))
```

## Recap

- `print()` has `sep` and `end`. `input()` always returns **text**.
- Convert with `int()`, `float()`, `str()` and `bool()`. `int("3.5")` fails, and `int(3.9)` is 3.
- A colon opens a block. Indent the block with **4 spaces**.
- Falsy values: `0`, `0.0`, `""`, `None`, `False`. Everything else is truthy.
- Built-ins save time: `len`, `abs`, `round`, `min`, `max`, `sum`.
