## Step 1: Create Variables

A **variable** is a name that points to a value. We write the name, then `=`, then the value. A value can be text, a whole number or a decimal number.

**Step 1: type this.** Make three variables: `language` with the text `Python`, `version` with `3.12`, and `students` with `25`. Print all three in one `print()`.

```python
language = "Python"
version = 3.12
students = 25
print(language, version, students)
```

## Step 2: Change a Variable

A name can be moved to a new value at any time. `score = score + 5` reads: *work out the old score plus 5, then put the label `score` on the answer.*

**Step 2: type this.** Set `score` to 10 and print it. Set `score` to 20 and print it. Then add 5 to `score` and print it.

```python
score = 10
print(score)
score = 20
print(score)
score = score + 5
print(score)
```

## Step 3: Check Naming Rules

Names must follow rules. `isidentifier()` is a text helper that tells us if a text is a valid name. `keyword.iskeyword()` tells us if a word is reserved by Python. You will learn text helpers in Module 03.

**Step 3: type this.** Import `keyword`. Check `student_name`, `2age` and `my-name` with `.isidentifier()`. Check whether `class` is a keyword.

```python
import keyword
print("student_name valid?", "student_name".isidentifier())
print("2age valid?", "2age".isidentifier())
print("my-name valid?", "my-name".isidentifier())
print("class is a keyword?", keyword.iskeyword("class"))
```

## Step 4: Case Matters

Python treats `age` and `Age` as **two different names**. This is called being *case-sensitive*. Stick to small letters with underscores (`snake_case`) for variables.

**Step 4: type this.** Create `age` with 25 and `Age` with 30. Print both. Then print whether `age == Age`.

```python
age = 25
Age = 30
print(age, Age)
print(age == Age)
```

## Step 5: Constants by Convention

Python has no real constants. By habit, we write constant names in `UPPER_CASE`. Python will still let you change them, so the capital letters are a promise to other readers. For pi, use the `math` module.

**Step 5: type this.** Make `MAX_ATTEMPTS` equal to 3 and `PI` equal to 3.14159, and print both. Change `MAX_ATTEMPTS` to 5 and print it. Then import `math` and print `math.pi`.

```python
MAX_ATTEMPTS = 3
PI = 3.14159
print(MAX_ATTEMPTS, PI)
MAX_ATTEMPTS = 5
print("Python allowed this change:", MAX_ATTEMPTS)
import math
print(math.pi)
```

## Step 6: Meet the Basic Types

Every value has a **type**. `type()` tells you which one. Notice that a name can later point to a value of another type, because the *value* has the type, not the name.

**Step 6: type this.** Make `count = 42`, `price = 19.99`, `city = "Chennai"` and `is_open = True`. Print the type of each in one `print()`. Then set `count` to the text `forty-two` and print its type.

```python
count = 42
price = 19.99
city = "Chennai"
is_open = True
print(type(count), type(price), type(city), type(is_open))
count = "forty-two"
print(type(count))
```

## Step 7: Meet None

`None` means *there is no value here*. It is not `0` and not empty text. To test for it, use `is None`. Its type is `NoneType`.

**Step 7: type this.** Set `result` to `None`. Print `result`, its type, and whether `result is None`. Then print whether `None == 0` and `None == ""`.

```python
result = None
print(result)
print(type(result))
print(result is None)
print(None == 0)
print(None == "")
```

## Step 8: Inspect with type and id

`id()` returns a number that identifies a value. If two names have the same `id`, they point to **the same value**. The exact number changes from run to run, so we only *compare* ids.

**Step 8: type this.** Set `x = 10` and `y = x`. Print whether `id(x) == id(y)`. Then set `y = y + 1`, print `x` and `y`, and print whether the ids are still equal.

```python
x = 10
y = x
print(id(x) == id(y))
y = y + 1
print(x, y)
print(id(x) == id(y))
```

## Recap

- A **variable** is a name that points to a value. `=` attaches the name.
- Names use letters, digits and `_`, cannot start with a digit, and are **case-sensitive**.
- Constants use `UPPER_CASE`. It is a convention, and Python does not enforce it.
- Basic types: `int`, `float`, `str`, `bool`. `None` means *no value*.
- `type()` shows the kind of value. `id()` shows its identity.
