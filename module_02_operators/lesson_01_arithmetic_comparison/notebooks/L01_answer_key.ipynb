## Step 1: Basic Arithmetic

Python can add, subtract, multiply and divide. `+`, `-` and `*` work as you expect. Note that `/` **always** gives a float.

**Step 1: type this.** Print `7 + 3`, `7 - 3`, `7 * 3` and `7 / 2`.

```python
print(7 + 3)
print(7 - 3)
print(7 * 3)
print(7 / 2)
```

## Step 2: Floor Division and Remainder

`//` divides and **rounds down**. `%` gives the **remainder**. `**` means power. Look carefully at the negative numbers. Floor division rounds towards the *smaller* number.

**Step 2: type this.** Print `7 // 2`, `7 % 2`, `-7 // 2`, `-7 % 2` and `2 ** 10`.

```python
print(7 // 2)
print(7 % 2)
print(-7 // 2)
print(-7 % 2)
print(2 ** 10)
```

## Step 3: Check the Divide Rule

Floor division and remainder always fit together: `a == (a // b) * b + a % b`. Python also has `divmod()`, which gives both answers at once as a pair.

**Step 3: type this.** Set `a = 17` and `b = 5`. Print `a // b` and `a % b`. Print whether the rule holds. Then print `divmod(a, b)`.

```python
a = 17
b = 5
print(a // b, a % b)
print((a // b) * b + a % b == a)
print(divmod(a, b))
```

## Step 4: Mixing Integers and Floats

The result type depends on the operator and the operands. `/` always gives a `float`. `//` gives an `int` if both sides are `int`, and a `float` if one side is a `float`. A negative power gives a float.

**Step 4: type this.** Print the type of `6 / 3`, `6 // 3` and `6 // 3.0`. Then print `2 ** -1`.

```python
print(type(6 / 3))
print(type(6 // 3))
print(type(6 // 3.0))
print(2 ** -1)
```

## Step 5: Remainder in Real Life

The remainder is very handy. `n % 2 == 0` tells you whether `n` is **even**. And `hours % 24` wraps a number of hours back onto the clock.

**Step 5: type this.** Set `number = 14` and print whether `number % 2 == 0`. Set `hours = 26` and print `hours % 24`.

```python
number = 14
print(number % 2 == 0)
hours = 26
print(hours % 24)
```

## Step 6: Assignment Operators

`x += 5` is a short way to write `x = x + 5`. Every arithmetic operator has this form. Notice that `/=` turns the value into a `float`.

**Step 6: type this.** Set `total = 10`. Then use `+= 5`, `-= 3`, `*= 2`, `//= 5`, `**= 2` and `/= 4`, printing `total` after each one.

```python
total = 10
total += 5
print(total)
total -= 3
print(total)
total *= 2
print(total)
total //= 5
print(total)
total **= 2
print(total)
total /= 4
print(total)
```

## Step 7: Comparison Operators

A comparison asks a question and gives `True` or `False`. Remember: `=` stores a value, `==` asks a question.

**Step 7: type this.** Print `5 == 5`, `5 != 3`, `3 > 5`, `3 < 5`, `5 >= 5` and `3 <= 2`.

```python
print(5 == 5)
print(5 != 3)
print(3 > 5)
print(3 < 5)
print(5 >= 5)
print(3 <= 2)
```

## Step 8: Chained and Text Comparisons

Python lets you chain comparisons: `18 <= age < 60` reads like maths. Text can be compared too. It follows the code order of characters, so capital letters come **before** small letters. And `5 == "5"` is `False`, because a number is not text.

**Step 8: type this.** Set `age = 25`. Print `18 <= age < 60`. Then print `"apple" < "banana"`, `"Zebra" < "apple"`, `5 == 5.0` and `5 == "5"`.

```python
age = 25
print(18 <= age < 60)
print("apple" < "banana")
print("Zebra" < "apple")
print(5 == 5.0)
print(5 == "5")
```

## Step 9: Decimals and isclose

Most decimals are stored as very close guesses, so `0.1 + 0.2` is not exactly `0.3`. Never compare floats with `==`. Use `math.isclose()`, or round first.

**Step 9: type this.** Print `0.1 + 0.2`. Print whether it equals `0.3`. Import `math` and print `math.isclose(0.1 + 0.2, 0.3)`. Then print whether `round(0.1 + 0.2, 2) == 0.3`.

```python
print(0.1 + 0.2)
print(0.1 + 0.2 == 0.3)
import math
print(math.isclose(0.1 + 0.2, 0.3))
print(round(0.1 + 0.2, 2) == 0.3)
```

## Step 10: Precedence

Python follows an order: brackets, `**`, unary minus, `*` `/` `//` `%`, `+` `-`, then comparisons. Use brackets whenever you are not sure. Note that `**` goes right to left.

**Step 10: type this.** Print `2 + 3 * 4`, `(2 + 3) * 4`, `-2 ** 2`, `(-2) ** 2`, `2 ** 3 ** 2`, `10 - 4 - 3` and `10 > 5 + 2`.

```python
print(2 + 3 * 4)
print((2 + 3) * 4)
print(-2 ** 2)
print((-2) ** 2)
print(2 ** 3 ** 2)
print(10 - 4 - 3)
print(10 > 5 + 2)
```

## Recap

- `/` gives a float. `//` rounds down (even for negatives). `%` is the remainder. `**` is power.
- `a == (a // b) * b + a % b` always holds. `n % 2 == 0` tests for even numbers.
- `+=`, `-=`, `*=`, `/=`, `//=`, `%=` and `**=` update a variable.
- `==` asks, `=` stores. Chain comparisons like `18 <= age < 60`. A number is never equal to text.
- Never compare floats with `==`. Use `math.isclose()` or `round()`.
- Precedence: brackets, `**`, unary minus, `* / // %`, `+ -`, comparisons. Add brackets when unsure.
