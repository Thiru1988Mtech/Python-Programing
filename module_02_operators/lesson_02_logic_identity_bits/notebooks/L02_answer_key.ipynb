## Step 1: and, or, not

`and` needs **both** sides to be `True`. `or` needs **at least one**. `not` flips the answer. You can combine them with comparisons.

**Step 1: type this.** Print `True and False`, `True or False` and `not True`. Then print `5 > 3 and 2 < 1` and `5 > 3 or 2 < 1`.

```python
print(True and False)
print(True or False)
print(not True)
print(5 > 3 and 2 < 1)
print(5 > 3 or 2 < 1)
```

## Step 2: Build Truth Tables

Let's print every case for `and` and `or`. The four pairs are `True True`, `True False`, `False True` and `False False`. Compare with the picture on the slide.

**Step 2: type this.** Print the four results of `and` on one line. Print the four results of `or` on the next line. Use the pairs in the order above.

```python
print(True and True, True and False, False and True, False and False)
print(True or True, True or False, False or True, False or False)
```

## Step 3: Conditions in Real Life

Real conditions mix comparisons and logic. Name the parts with variables so the code reads like a sentence.

**Step 3: type this.** Set `age = 25` and `has_id = True`. Print `age >= 18 and has_id`. Print `age < 13 or age > 60`. Print `not has_id`.

```python
age = 25
has_id = True
print(age >= 18 and has_id)
print(age < 13 or age > 60)
print(not has_id)
```

## Step 4: Short-Circuit Logic

`and` and `or` do not always return `True` or `False`. They return **one of their values**. `or` gives the first truthy value. `and` gives the first falsy value, or the last value. Python stops as soon as it knows the answer.

**Step 4: type this.** Print `0 or "default"`, `"" or "guest"`, `5 and 10` and `0 and 10`. Then set `name` to empty text and print `name or "Anonymous"`.

```python
print(0 or "default")
print("" or "guest")
print(5 and 10)
print(0 and 10)
name = ""
print(name or "Anonymous")
```

## Step 5: Order of not, and, or

The order is `not` first, then `and`, then `or`. Think of `and` as multiply and `or` as add. Brackets change the order.

**Step 5: type this.** Print `True or False and False`, `(True or False) and False`, `not True or True` and `not (True or True)`.

```python
print(True or False and False)
print((True or False) and False)
print(not True or True)
print(not (True or True))
```

## Step 6: == versus is

`==` asks whether two values are **equal**. `is` asks whether two names point to the **same object**. Two lists with the same items are equal, but they are two separate objects. We meet lists properly in Module 04.

**Step 6: type this.** Make `a = [1, 2]`, `b = [1, 2]` and `c = a`. Print `a == b`, `a is b`, `a is c` and whether `id(a) == id(c)`.

```python
a = [1, 2]
b = [1, 2]
c = a
print(a == b)
print(a is b)
print(a is c)
print(id(a) == id(c))
```

## Step 7: Checking for None

The correct way to test for `None` is `is None` or `is not None`. There is only one `None` object in Python, so `is` is exact and fast.

**Step 7: type this.** Set `result = None`. Print `result is None` and `result is not None`. Then set `result = 42` and print `result is not None`.

```python
result = None
print(result is None)
print(result is not None)
result = 42
print(result is not None)
```

## Step 8: Membership with in

`in` asks whether a piece is inside a group. `not in` asks whether it is missing. They work on text and on lists. Text checks are case-sensitive.

**Step 8: type this.** Print `"py" in "python"`, `"Py" in "python"`, `3 in [1, 2, 3]` and `4 not in [1, 2, 3]`. Then set `letter = "e"` and print whether it is in `"aeiou"`.

```python
print("py" in "python")
print("Py" in "python")
print(3 in [1, 2, 3])
print(4 not in [1, 2, 3])
letter = "e"
print(letter in "aeiou")
```

## Step 9: Bitwise Operators

Bitwise operators work on the bits of whole numbers. `&` keeps bits that are in both. `|` keeps bits that are in either. `^` keeps bits that differ. `~` flips all bits. `<<` and `>>` shift the bits. Shifting left by one doubles a number.

**Step 9: type this.** Print `5 & 3`, `5 | 3`, `5 ^ 3`, `~5`, `5 << 1` and `5 >> 1`. Then print whether `5 << 1 == 5 * 2`.

```python
print(5 & 3)
print(5 | 3)
print(5 ^ 3)
print(~5)
print(5 << 1)
print(5 >> 1)
print(5 << 1 == 5 * 2)
```

## Step 10: See the Bits with bin()

`bin()` shows the bits of a number. `0b` only means *binary*. `format(n, "04b")` shows 4 digits, padded with zeros. The last bit tells you if a number is odd: `n & 1`.

**Step 10: type this.** Print `bin(5)` and `bin(3)`. Print `bin()` of `5 & 3`, `5 | 3` and `5 ^ 3`. Print `format(5, "04b")`. Then print `7 & 1` and `6 & 1`.

```python
print(bin(5), bin(3))
print(bin(5 & 3), bin(5 | 3), bin(5 ^ 3))
print(format(5, "04b"))
print(7 & 1, 6 & 1)
```

## Recap

- `and` needs both, `or` needs one, `not` flips. Order: `not`, `and`, `or`.
- `and` and `or` return one of their values: `name or "Anonymous"` gives a default.
- `==` compares values. `is` compares objects. Use `is None` and `is not None`.
- `in` and `not in` test membership in text and in lists. Text checks are case-sensitive.
- `&`, `|`, `^`, `~`, `<<`, `>>` work on bits. Use `bin()` to look at them. `n & 1` tests odd or even.
