## Step 1: The if Statement

`if condition:` is followed by an **indented block**. The block runs only when the condition is `True`. Lines that are not indented always run.

**Step 1: type this.** Set `age` to `20`. Write an `if` that prints `adult` when `age >= 18`. Add one unindented `print("checked")`. Then set `age` to `15` and run the same code again in a new cell part.

```python
age = 20
if age >= 18:
    print("adult")
print("checked")
age = 15
if age >= 18:
    print("adult")
print("checked")
```

## Step 2: if and else

`else:` runs when the condition is `False`. Exactly one of the two blocks runs. `else` has no condition of its own.

**Step 2: type this.** Set `temp` to `35`. If `temp > 30`, print `hot`, else print `pleasant`. Then set `temp` to `22` and repeat.

```python
temp = 35
if temp > 30:
    print("hot")
else:
    print("pleasant")
temp = 22
if temp > 30:
    print("hot")
else:
    print("pleasant")
```

## Step 3: Build Conditions

Combine conditions with `and`, `or` and `not`. Use `in` to test membership. Brackets make the order clear.

**Step 3: type this.** Set `age = 25`, `has_ticket = True`, `day = "sat"`. Print `age >= 18 and has_ticket`, `day in ("sat", "sun")`, `not has_ticket` and `age < 13 or age > 60`. Then use an `if` with the first condition.

```python
age = 25
has_ticket = True
day = "sat"
print(age >= 18 and has_ticket)
print(day in ("sat", "sun"))
print(not has_ticket)
print(age < 13 or age > 60)
if age >= 18 and has_ticket and day in ("sat", "sun"):
    print("weekend entry")
```

## Step 4: Truthy and Falsy

Any value can be a condition. **Falsy** values are `0`, `0.0`, `""`, `[]`, `{}`, `set()` and `None`. Everything else is **truthy**. `bool()` shows how Python sees a value.

**Step 4: type this.** Print `bool(0)`, `bool("")`, `bool([])`, `bool(None)`, `bool("0")` and `bool([0])`. Then set `name = ""` and use `if name:` with an `else` to print `anonymous` or the name.

```python
print(bool(0), bool(""), bool([]), bool(None))
print(bool("0"), bool([0]), bool(-1))
name = ""
if name:
    print("Hello", name)
else:
    print("Hello, anonymous")
```

## Step 5: The elif Ladder

`elif` adds more conditions. Python checks from the top, and runs the **first** block whose condition is `True`. The rest are skipped. `else` is last and optional.

**Step 5: type this.** Set `marks = 72`. Print `A` for 90 or more, `B` for 70 or more, `C` for 50 or more, otherwise `F`. Wrap it so you can run it for `marks` in `95`, `72`, `55` and `30` by changing the value and running each time (write the four tests one after another).

```python
def grade(marks):
    if marks >= 90:
        return "A"
    elif marks >= 70:
        return "B"
    elif marks >= 50:
        return "C"
    else:
        return "F"

for m in (95, 72, 55, 30):
    print(m, grade(m))
```

## Step 6: Order and Separate ifs

Order matters in `elif`. Two separate `if` statements are checked independently, so both blocks can run. Compare the same marks with `elif` and with separate `if` lines.

**Step 6: type this.** Set `marks = 95`. Write a **wrong** ladder that tests `>= 50` first. Then write separate `if` lines for `>= 90` and `>= 50`. Then the correct `elif` ladder. Print the result of each.

```python
marks = 95
if marks >= 50:
    print("wrong ladder: C")
elif marks >= 90:
    print("wrong ladder: A")
if marks >= 90:
    print("separate if: A")
if marks >= 50:
    print("separate if: C")
if marks >= 90:
    print("elif ladder: A")
elif marks >= 50:
    print("elif ladder: C")
```

## Step 7: Nested Decisions

A decision inside a decision is a **nested** `if`. Each level adds four spaces. Flatten with `and` when the inner test is a simple extra condition.

**Step 7: type this.** Set `age = 20` and `has_id = True`. Write a nested `if` that prints `enter`, `show ID` or `too young`. Then write the same logic with `elif` and `and`, and print which one is shorter to read.

```python
age = 20
has_id = True
if age >= 18:
    if has_id:
        print("enter")
    else:
        print("show ID")
else:
    print("too young")
if age < 18:
    print("too young")
elif has_id:
    print("enter")
else:
    print("show ID")
```

## Step 8: Conditional Expression

`value_if_true if condition else value_if_false` is an **expression**, so it gives a value. Use it for short, simple choices, such as a label or a default.

**Step 8: type this.** Set `age = 20` and make `label`. Print it. Print `"odd" if 7 % 2 else "even"`. Make `bigger` using `if` and `else` on `3` and `5`. Use it inside an f-string: `f"{n} is {'odd' if n % 2 else 'even'}"` for `n = 4`.

```python
age = 20
label = "adult" if age >= 18 else "minor"
print(label)
print("odd" if 7 % 2 else "even")
bigger = 3 if 3 > 5 else 5
print(bigger)
n = 4
print(f"{n} is {'odd' if n % 2 else 'even'}")
```

## Step 9: Bonus: match

`match` compares one value with several patterns. It needs **Python 3.10 or newer**. `case _:` is the default. `|` joins several values in one case.

**Step 9: type this.** Write `describe(code)` with `match`: `200` gives `ok`, `404` or `410` gives `gone`, anything else gives `other`. Print the result for `200`, `404` and `500`.

```python
def describe(code):
    match code:
        case 200:
            return "ok"
        case 404 | 410:
            return "gone"
        case _:
            return "other"

print(describe(200))
print(describe(404))
print(describe(500))
```

## Step 10: Mini Practice: Ticket Price

Let's use everything. A cinema charges by age: under 5 is free, under 18 is 100, 60 or more is 80, and everyone else pays 150. On weekends, add 20. Use `elif`, a conditional expression and an f-string.

**Step 10: type this.** Make a function `price(age, weekend)`. Print the price for `(4, False)`, `(10, True)`, `(30, False)` and `(65, True)`.

```python
def price(age, weekend):
    if age < 5:
        base = 0
    elif age < 18:
        base = 100
    elif age >= 60:
        base = 80
    else:
        base = 150
    extra = 20 if weekend and base > 0 else 0
    return base + extra

for age, weekend in [(4, False), (10, True), (30, False), (65, True)]:
    print(f"age {age:>2}, weekend={weekend!s:<5} -> {price(age, weekend)}")
```

## Recap

- `if condition:` plus an indented block. `else:` is the False case. One block runs.
- Conditions can be comparisons, `and` / `or` / `not`, `in`, or any truthy or falsy value.
- `elif` is a ladder: the first True wins, so put the strictest test first. Separate `if` lines are independent.
- Nest sparingly. Flatten with `and` or with `elif`.
- `a if cond else b` gives a value. `match` compares one value with patterns (3.10+).
