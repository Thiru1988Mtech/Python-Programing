## Step 1: break: Stop Early

`break` ends the loop immediately. The program jumps to the first line after the loop. It is usually inside an `if`.

**Step 1: type this.** Loop over `[4, 7, 9, 12]`. For each `n`, if `n > 8`, print `found` and the number, and `break`. Otherwise print `checked`. Print `end` after the loop.

```python
for n in [4, 7, 9, 12]:
    if n > 8:
        print("found", n)
        break
    print("checked", n)
print("end")
```

## Step 2: continue: Skip a Round

`continue` skips the rest of the current round and goes to the next one. The loop itself keeps going.

**Step 2: type this.** Loop over `range(1, 6)`. Use `continue` to skip even numbers, and print the odd ones. Then loop over `["a", "", "b", ""]` and skip empty items, collecting the others in `clean`.

```python
for n in range(1, 6):
    if n % 2 == 0:
        continue
    print(n)
clean = []
for text in ["a", "", "b", ""]:
    if not text:
        continue
    clean.append(text)
print(clean)
```

## Step 3: Search with break in while

`break` also works in a `while` loop. A `while True:` loop runs until a `break`. Always make sure that a `break` can really happen.

**Step 3: type this.** Use `while True` with a counter `n`. Add 1 to `n` each round, and `break` when `n * n > 50`. Print the first `n` whose square is above 50.

```python
n = 0
while True:
    n += 1
    if n * n > 50:
        break
print(n, n * n)
```

## Step 4: pass: A Placeholder

Python needs at least one line in every block. `pass` is a line that does nothing. Use it while you are planning code you will write later.

**Step 4: type this.** Write an `if` with `pass`, a `for` loop with `pass` and print `ok` after them. Then write `if` and `else` where only the `else` does something.

```python
x = 5
if x > 10:
    pass
for i in range(3):
    pass
print("ok")
if x > 3:
    pass
else:
    print("small")
print("end")
```

## Step 5: for-else

`else` on a loop runs when the loop **finishes without** a `break`. It is a neat way to say: the search failed. If a `break` happens, the `else` is skipped.

**Step 5: type this.** Search `[3, 5, 7]` for an even number. Use `break` in the loop and an `else` for the not-found case. Then search `[3, 4, 7]`.

```python
for nums in ([3, 5, 7], [3, 4, 7]):
    for n in nums:
        if n % 2 == 0:
            print("even", n)
            break
    else:
        print("no even in", nums)
```

## Step 6: Nested Loops: a Grid

An inner loop runs completely for every round of the outer loop. Build a row of text in the inner loop, and print it after the inner loop ends.

**Step 6: type this.** Print a 3 by 4 multiplication grid with rows `r` from `1` to `3` and columns `c` from `1` to `4`. Use `f"{r * c:>3}"` for each cell and print each row. Then print every pair of `"ab"` and `"12"`.

```python
for r in range(1, 4):
    row = ""
    for c in range(1, 5):
        row += f"{r * c:>3}"
    print(row)
for a in "ab":
    for b in "12":
        print(a + b, end=" ")
print()
```

## Step 7: Nested Loops and break

`break` leaves only the **inner** loop. The outer loop carries on with its next round. Use it to find the first match in each row.

**Step 7: type this.** Given `rows = [[1, 3, 5], [2, 4, 6], [7, 9, 8]]`, find the first even number in each row. Use `break` in the inner loop. Print the row index and the number, or `none` if the row has no even number.

```python
rows = [[1, 3, 5], [2, 4, 6], [7, 9, 8], [1, 1, 1]]
for i, row in enumerate(rows):
    first_even = None
    for n in row:
        if n % 2 == 0:
            first_even = n
            break
    print(i, first_even if first_even else "none")
```

## Step 8: Safe Loops

Do **not** remove items from a list while looping over it. Python skips items. Loop over a **copy**, or build a **new** list. A guard counter can stop a `while` loop that may run too long.

**Step 8: type this.** Remove all even numbers from `[1, 2, 4, 5, 6]`. First show the wrong way. Then the safe way with a copy `nums[:]`. Then use a guard counter `tries` in a `while True` loop that stops at 5.

```python
nums = [1, 2, 4, 5, 6]
for n in nums:
    if n % 2 == 0:
        nums.remove(n)
print("wrong", nums)
nums = [1, 2, 4, 5, 6]
for n in nums[:]:
    if n % 2 == 0:
        nums.remove(n)
print("safe", nums)
tries = 0
while True:
    tries += 1
    if tries >= 5:
        break
print("tries", tries)
```

## Step 9: Mini Practice: Guessing Game

Let's use everything. A guessing game: the secret is `42`. We feed a list of guesses. For each guess, print `too low`, `too high`, or `correct` and stop. If the guesses run out, print `out of guesses` using the loop `else`. Skip guesses that are not numbers.

**Step 9: type this.** Make `guesses = [50, "x", 30, 42, 10]`. Loop with `for`, skip non-numbers with `continue`, and `break` on the correct guess. Add an `else` for the case of no correct guess. Count the tries.

```python
secret = 42
guesses = [50, "x", 30, 42, 10]
tries = 0
for guess in guesses:
    if not isinstance(guess, int):
        continue
    tries += 1
    if guess < secret:
        print(guess, "too low")
    elif guess > secret:
        print(guess, "too high")
    else:
        print(guess, "correct in", tries, "tries")
        break
else:
    print("out of guesses")
```

## Recap

- `break` leaves the loop. `continue` skips one round. Use both with an `if`.
- `pass` is a placeholder. The `else` of a loop runs when there was no `break`.
- In nested loops, the inner loop runs fully for each outer round. `break` leaves only the inner loop.
- Do not change a list while you loop over it. Loop over a copy: `for n in nums[:]`.
- Check where loops start and stop. Give `while True` an exit.
