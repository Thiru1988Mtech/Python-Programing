## Step 1: Loop over a List

`for item in collection:` runs the indented body once for each item. After the last item, the loop ends and the next unindented line runs.

**Step 1: type this.** Make `fruits` with `apple`, `banana`, `cherry`. Loop over it and print each fruit in capital letters. After the loop, print `done`.

```python
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit.upper())
print("done")
```

## Step 2: Loop over Text and Accumulate

A string is a collection of characters, so `for ch in text:` works. An **accumulator** is a variable that you start before the loop and update inside it, such as a running total or count.

**Step 2: type this.** Loop over `"banana"` and count the letter `a` with a counter `count`. Then add up `[4, 8, 15]` into `total`. Print `count` and `total`.

```python
count = 0
for ch in "banana":
    if ch == "a":
        count += 1
print(count)
total = 0
for n in [4, 8, 15]:
    total += n
print(total)
```

## Step 3: range() Forms

`range(stop)`, `range(start, stop)` and `range(start, stop, step)`. The stop is **never** included. `list()` shows the numbers. Use a negative step to count down.

**Step 3: type this.** Print `list(range(5))`, `list(range(1, 6))`, `list(range(0, 10, 2))`, `list(range(5, 0, -1))` and `len(range(100))`.

```python
print(list(range(5)))
print(list(range(1, 6)))
print(list(range(0, 10, 2)))
print(list(range(5, 0, -1)))
print(len(range(100)))
```

## Step 4: Loop with range

`for i in range(n):` repeats a block `n` times. Use `range(1, n + 1)` to count from 1 to `n`. You can also use `range(len(items))` to get the positions.

**Step 4: type this.** Print `hello` three times using `range(3)`. Print the squares of `1` to `5` with `range(1, 6)`. Then use `range(len(colors))` on `["red", "green", "blue"]` to print each position and colour.

```python
for i in range(3):
    print("hello", i)
for n in range(1, 6):
    print(n, n ** 2)
colors = ["red", "green", "blue"]
for i in range(len(colors)):
    print(i, colors[i])
```

## Step 5: The while Loop

`while condition:` checks the condition **before** every round. The body must change something so that the condition becomes `False`. Otherwise the loop is **infinite**.

**Step 5: type this.** Make `count = 3`. While `count > 0`, print it and subtract 1. Then print `go!`. Next, double `x = 1` while `x < 100` and print each value.

```python
count = 3
while count > 0:
    print(count)
    count -= 1
print("go!")
x = 1
while x < 100:
    x *= 2
    print(x)
```

## Step 6: while with a Counter and a List

A `while` loop can walk a list with an index, or wait for a condition that depends on the data. Here we keep adding Fibonacci numbers while the next one is below 50.

**Step 6: type this.** Make `fib = [1, 1]`. While the last two numbers add up to less than `50`, append their sum. Print `fib`. Then use a `while` with an index `i` to print each item of `["a", "b", "c"]`.

```python
fib = [1, 1]
while fib[-1] + fib[-2] < 50:
    fib.append(fib[-1] + fib[-2])
print(fib)
items = ["a", "b", "c"]
i = 0
while i < len(items):
    print(i, items[i])
    i += 1
```

## Step 7: Loop over Dictionaries and Sets

Looping over a dictionary gives the **keys**. `.values()` gives values and `.items()` gives `(key, value)` pairs that you can unpack. Sets have no order, so use `sorted()` when order matters.

**Step 7: type this.** Make `prices` with `tea` 10, `cake` 40, `juice` 25. Loop over keys, then over `.items()`, and compute `total` from `.values()`. Then loop over `sorted({3, 1, 2})`.

```python
prices = {"tea": 10, "cake": 40, "juice": 25}
for name in prices:
    print(name)
for name, cost in prices.items():
    print(f"{name:<6} {cost:>3}")
total = 0
for cost in prices.values():
    total += cost
print(total)
for x in sorted({3, 1, 2}):
    print(x)
```

## Step 8: enumerate and Building Lists

`enumerate(items)` gives `(index, item)` pairs. `enumerate(items, start=1)` starts counting at 1. To build a new list, start with `[]` and `append` inside the loop.

**Step 8: type this.** Loop over `["red", "green", "blue"]` with `enumerate(..., start=1)`. Then build `evens` with the even numbers from `range(10)`, and `squares` of `[1, 2, 3]`. Print both.

```python
for pos, color in enumerate(["red", "green", "blue"], start=1):
    print(pos, color)
evens = []
for n in range(10):
    if n % 2 == 0:
        evens.append(n)
print(evens)
squares = []
for n in [1, 2, 3]:
    squares.append(n * n)
print(squares)
```

## Step 9: Mini Practice: Times Table

Let's use everything. Print the times table of 7 from 1 to 10 with a nice format. Then use a `while` to find the first power of 2 above 1000, and a `for` to add the numbers 1 to 100.

**Step 9: type this.** Loop `n` over `range(1, 11)` and print `7 x n = result` aligned. Find `power` with `while`. Print `sum` of 1 to 100 using a loop with `total`.

```python
for n in range(1, 11):
    print(f"7 x {n:>2} = {7 * n:>2}")
power = 1
while power <= 1000:
    power *= 2
print(power)
total = 0
for n in range(1, 101):
    total += n
print(total)
```

## Recap

- `for item in collection:` runs the body once per item. Set an accumulator before the loop.
- `range(start, stop, step)`: the stop is never included. `range(n)` repeats `n` times.
- `while condition:` checks first. The body must change something, or it never ends.
- A dictionary loop gives keys. Use `.items()` for pairs, `.values()` for values and `enumerate()` for positions.
- Build a list with `[]` and `.append()`. Prefer `for` over `while` when you can.
